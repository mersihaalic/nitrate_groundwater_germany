```python
import pandas as pd
```

```python
excel_file = pd.ExcelFile("../data/raw/Nitrate Dataset Germany.xlsx")

excel_file.sheet_names
```

```python
df_stations = pd.read_excel(
    "../data/raw/Nitrate Dataset Germany.xlsx",
    sheet_name="GW_Messstellen"
)

df_nitrate = pd.read_excel(
    "../data/raw/Nitrate Dataset Germany.xlsx",
    sheet_name="GW_Nitrat_MW"
)

df_period = pd.read_excel(
    "../data/raw/Nitrate Dataset Germany.xlsx",
    sheet_name="GW_Werte_Zeitraum"
)
```

```python
print("Stations:", df_stations.shape)
print("Nitrate:", df_nitrate.shape)
print("Period:", df_period.shape)
```

```python
print("STATIONS COLUMNS:")
print(df_stations.columns.tolist())

print("\nNITRATE COLUMNS:")
print(df_nitrate.columns.tolist())

print("\nPERIOD COLUMNS:")
print(df_period.columns.tolist())
```

```python
print("Unique stations in df_stations:", df_stations["Messstellencode"].nunique())
print("Rows in df_stations:", len(df_stations))
print("Duplicates in df_stations:", df_stations["Messstellencode"].duplicated().sum())

print("\nUnique stations in df_nitrate:", df_nitrate["Messstellencode"].nunique())
print("Rows in df_nitrate:", len(df_nitrate))

print("\nUnique stations in df_period:", df_period["Messstellencode"].nunique())
print("Rows in df_period:", len(df_period))
```

```python
df_master = pd.merge(
    df_nitrate,
    df_stations,
    on="Messstellencode",
    how="left"
)

df_master.head()
```

```python
print("Rows before merge:", len(df_nitrate))
print("Rows after merge:", len(df_master))

print("\nMissing station names:", df_master["Messstellenname"].isna().sum())
print("Missing groundwater codes:", df_master["Wasserkörpercode"].isna().sum())
```

```python
df_master.to_csv(
    "../data/processed/groundwater_nitrate_annual.csv",
    index=False
)
```
