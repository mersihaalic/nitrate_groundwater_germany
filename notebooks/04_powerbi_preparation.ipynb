```python
import pandas as pd

df = pd.read_csv("../data/processed/groundwater_nitrate_annual.csv")

df.head()
```

```python
df.info()
```

```python
df_powerbi = df.rename(columns={
    "Messstellencode": "StationCode",
    "Jahr": "Year",
    "Anzahl Messungen": "NumberOfMeasurements",
    "Jahresmittelwert in mg Nitrat pro Liter": "Nitrate_mg_L",
    "Messstellenname": "StationName",
    "Wasserkörpercode": "GroundwaterBodyCode",
    "Wasserkörpername": "GroundwaterBodyName",
    "Probenahmetiefe in m": "SamplingDepth_m",
    "Länge in Grad (ETRS89)": "Longitude",
    "Breite in Grad (ETRS89)": "Latitude"
})

df_powerbi.head()
```

```python
df
```

```python
df_powerbi
```

```python
df_powerbi.columns
```

```python
df_powerbi["NitrateStatus"] = df_powerbi["Nitrate_mg_L"].apply(
    lambda x: "Above 50 mg/L" if x > 50 else "At or below 50 mg/L"
)

df_powerbi[["Nitrate_mg_L", "NitrateStatus"]].head(10)
```

```python
df_powerbi["NitrateStatus"].value_counts()
```

```python
df_powerbi["NitrateCategory"] = pd.cut(
    df_powerbi["Nitrate_mg_L"],
    bins=[0, 25, 50, 100, float("inf")],
    labels=[
        "Low (0-25 mg/L)",
        "Moderate (25-50 mg/L)",
        "High (50-100 mg/L)",
        "Very high (>100 mg/L)"
    ],
    include_lowest=True
)

df_powerbi["NitrateCategory"].value_counts().sort_index()
```

```python
df_powerbi.isna().sum()
```

```python
import numpy as np

df_powerbi["StationName"] = df_powerbi["StationName"].replace(
    ["", " ", "NA", "N/A"],
    np.nan
)
```

```python
df_powerbi.isna().sum()
```

```python
df_powerbi["StateCode"] = df_powerbi["StationCode"].str.split("_").str[0]

df_powerbi["StateCode"].value_counts().sort_index()
```

```python
state_mapping = {
    "BB": "Brandenburg",
    "BW": "Baden-Württemberg",
    "BY": "Bayern",
    "HB": "Bremen",
    "HE": "Hessen",
    "HH": "Hamburg",
    "MV": "Mecklenburg-Vorpommern",
    "NI": "Niedersachsen",
    "NW": "Nordrhein-Westfalen",
    "RP": "Rheinland-Pfalz",
    "SH": "Schleswig-Holstein",
    "SL": "Saarland",
    "SN": "Sachsen",
    "ST": "Sachsen-Anhalt",
    "TH": "Thüringen"
}

df_powerbi["FederalState"] = df_powerbi["StateCode"].map(state_mapping)
```

```python
df_powerbi[
    ["StationCode", "StateCode", "FederalState"]
].head(10)
```

```python
df_powerbi["FederalState"].isna().sum()
```

```python
df_powerbi["FederalState"].value_counts()
```

```python
print("Shape:", df_powerbi.shape)

print("\nMissing values:")
print(df_powerbi.isna().sum())

print("\nDuplicate Station-Year rows:")
print(
    df_powerbi.duplicated(
        subset=["StationCode", "Year"]
    ).sum()
)

print("\nYear range:")
print(df_powerbi["Year"].min(), "-", df_powerbi["Year"].max())

print("\nNitrate range:")
print(
    df_powerbi["Nitrate_mg_L"].min(),
    "-",
    df_powerbi["Nitrate_mg_L"].max()
)

print("\nNumber of stations:")
print(df_powerbi["StationCode"].nunique())

print("\nNumber of federal states:")
print(df_powerbi["FederalState"].nunique())
```

```python
print("State codes in data:")
print(sorted(df_powerbi["StateCode"].unique()))

print("\nState codes in mapping:")
print(sorted(state_mapping.keys()))

print("\nMissing from mapping:")
print(set(df_powerbi["StateCode"].unique()) - set(state_mapping.keys()))
```

```python
df_powerbi[
    ["StateCode", "FederalState"]
].drop_duplicates().sort_values("StateCode")
```

```python
df_powerbi.to_csv(
    "../data/processed/powerbi_groundwater_nitrate.csv",
    index=False
)
```
